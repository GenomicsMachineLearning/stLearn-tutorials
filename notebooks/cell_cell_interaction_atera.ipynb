# Cell-Cell Interaction Analysis with Atera Data


### Environment setup

```python
import os
import platform
import random

# Make sure all the seeds are set
seed = 0
random.seed(seed)

# Only constrain threads on macOS where BLAS/numba deadlocks are common.
# Must run before any numpy/scanpy import.
if platform.system() == "Darwin":
    os.environ["OPENBLAS_NUM_THREADS"] = "1"
    os.environ["MKL_NUM_THREADS"] = "1"
    os.environ["NUMBA_NUM_THREADS"] = "1"
    n_cpus = 1
else:
    n_cpus = None
```

```python
import stlearn
import pathlib
import matplotlib.pyplot as plt
import numpy as np

stlearn.settings.set_figure_params(dpi=120)
np.random.seed(seed)

# Ignore all warnings
import warnings

warnings.filterwarnings("ignore")
```

### Loading the data
For this tutorial purpose, we don't perform the clustering. Please run the clustering method with the clustering tutorial or a part of spatial trajectory inference tutorial.

```python
# Setup download directory and get data
stlearn.settings.datasetdir = pathlib.Path.cwd().parent / "data"
sample_id = "WTA_Preview_FFPE_Breast_Cancer"
table_name = "table"
data_dir = stlearn.settings.datasetdir / sample_id
DEMO = False
```

```python
sdata = stlearn.datasets.atera_sge(sample_id=sample_id, aligned_images=True, include_hires_tiff=True)
adata = sdata.tables[table_name]
```

```python
sdata.tables
```

```python
# Crop the sample to the center quarter using a bounding box.
if DEMO:
    sdata = stlearn.io.store_checkpoint(
        data_dir / f"{sample_id}_{table_name}_demo.zarr", lambda:
        stlearn.io.crop_xenium(sdata, table_name)
    )
adata = sdata.tables[table_name]
print(f"{adata.n_obs:,} bins x {adata.n_vars:,} genes")
```

```python
# QC - Filter genes and cells with at least 10 counts, perform normalisation, HVG, PCA, NN and clustering.
clustered_name = f"{table_name}_clustered"

def cluster():
    adata = sdata.tables[table_name].copy()
    stlearn.pp.filter_genes(adata, min_counts=10)
    stlearn.pp.filter_cells(adata, min_counts=10)
    adata.layers["counts"] = adata.X.copy()
    stlearn.pp.normalize_total(adata)
    stlearn.pp.log1p(adata)
    stlearn.pp.highly_variable_genes(adata, n_top_genes=2000)
    stlearn.em.run_pca(adata, n_comps=50, random_state=seed)
    stlearn.pp.neighbors(adata, n_neighbors=25, use_rep="X_pca", random_state=seed)
    stlearn.tl.clustering.leiden(adata, resolution=1.05, random_state=seed)
    return adata

adata = stlearn.io.checkpoint(sdata, clustered_name, cluster)
```

```python
stlearn.pl.sdata_plot(sdata, "leiden", table_key=clustered_name)
```

```python
adata.X = adata.layers["counts"].copy()
stlearn.pp.normalize_total(adata)
```

## Gridding

Now performing the gridding. The resolution chosen here may effect the results. The higher resolution, the better this represents the single cell data but the longer the computation takes.

To summarise the gene expression across cells in a grid, we sum the library size normalised gene expression. Summing allows for representing the fact there are multiple cells in a given spot.

```python
### Calculating the number of grid spots we will generate
n_ = 125
print(f'{n_} by {n_} has this many spots:\n', n_ * n_)
```

```python
adata = stlearn.io.attach_spatial_sdata(adata, sdata, library_id=sample_id, microns_per_unit=stlearn.io.xenium_pixel_size(data_dir), image_key="he_image")
```

By providing 'use_label' to the function below, the cell type information is saved as deconvolution information
per spot, and also the dominant cell annotation. That way we can perform stLearn CCI with the cell type information!

```python
### Gridding.
grid_name = f"grid_{n_}x{n_}"
grid = stlearn.io.checkpoint(
    sdata, grid_name, lambda:
    stlearn.tl.cci.grid(adata, n_row=n_, n_col=n_, use_label='leiden', n_cpus=n_cpus)
    )
```

### Checking the gridding

Comparing the gridded data to the original, to make sure it makes sense.

It's recommend to visualise the dominant cell types per spot, in order to gauge whether the tissue structure is adequately maintained after gridding (i.e. to make sure it is not too low resolution!).

```python
fig, axes = plt.subplots(ncols=2, figsize=(20, 5))
stlearn.pl.cluster_plot(grid, use_label='leiden', size=10, ax=axes[0], show_plot=False, show_image=False)
stlearn.pl.cluster_plot(adata, use_label='leiden', ax=axes[1], show_plot=False, show_image=False)
axes[0].set_title(f'Grid leiden dominant spots')
axes[0].set_aspect('equal')
axes[1].set_title(f'Cell leiden labels')
axes[1].set_aspect('equal')
plt.show()
```

```python
groups = list(grid.obs['leiden'].cat.categories)
idc_cluster = groups[9]
dcis_cluster = groups[4]
stroma_cluster = groups[11]
for group in [idc_cluster, dcis_cluster, stroma_cluster]:
    fig, axes = plt.subplots(ncols=3, figsize=(20, 5))
    grid.obs['group'] = grid.uns['leiden'][group].values
    stlearn.pl.feat_plot(grid, feature='group', ax=axes[0], show_plot=False, vmax=1, show_color_bar=False, show_image=False)
    stlearn.pl.cluster_plot(grid, use_label='leiden', list_clusters=[group], ax=axes[1], show_plot=False, show_image=False)
    stlearn.pl.cluster_plot(adata, use_label='leiden', list_clusters=[group], ax=axes[2], show_plot=False, show_image=False)
    axes[0].set_title(f'Grid {group} proportions (max = 1)')
    axes[0].set_aspect('equal')
    axes[1].set_title(f'Grid {group} max spots')
    axes[1].set_aspect('equal')
    axes[2].set_title(f'Individual cell {group}')
    axes[2].set_aspect('equal')
    plt.show()
```

```python
fig, axes = plt.subplots(ncols=2, figsize=(20, 8))
stlearn.pl.gene_plot(grid, gene_symbols='CXCL12', ax=axes[0], show_color_bar=False, show_plot=False, show_image=False)
stlearn.pl.gene_plot(adata, gene_symbols='CXCL12', ax=axes[1], show_color_bar=False, show_plot=False, vmax=80,
                show_image=False)
axes[0].set_aspect('equal')
axes[1].set_aspect('equal')
axes[0].set_title(f'Grid CXCL12 expression')
axes[1].set_title(f'Cell CXLC12 expression')
plt.show()
```

## LR Permutation Test

Running the LR permutation test, to determine regions of high LR co-expression.

```python
# Loading the LR databases available within stlearn (from NATMI)
lrs = stlearn.tl.cci.load_lrs(['connectomeDB2020'], species='human')
print(len(lrs))
```

```python
# Running the analysis #
lr_name = f"{grid_name}_lr"
lr_results = stlearn.io.checkpoint(
    sdata, lr_name, compute=lambda:
    stlearn.tl.cci.run(
        grid, lrs,
        min_spots=20,  # Filter out any LR pairs with no scores for less than min_spots
        distance=250,  # None defaults to spot+immediate neighbours; distance=0 for within-spot mode
        n_pairs=1000,  # Number of random pairs to generate; low as example, recommend ~10,000
        n_cpus=n_cpus,  # Number of CPUs for parallel. If None, detects & use all available.
        random_state=seed
        )
    )
stlearn.io.save_table(sdata, grid_name)  # cci.run added the neighbours to grid
```

```python
stlearn.tl.cci.adj_pvals(lr_results, correct_axis='spot', pval_adj_cutoff=0.05, adj_method='fdr_bh')
```

```python
lr_info = lr_results.var  # A dataframe detailing the LR pairs ranked by number of significant spots.
lr_info
```

```python
# Showing the rankings of the LR from a global and local perspective.
# Ranking based on number of significant hotspots.
stlearn.pl.lr_summary(lr_results, n_top=500)
stlearn.pl.lr_summary(lr_results, n_top=50, figsize=(10, 3))
```

## Predicting Significant Cell-Cell Interactions

For this analysis, we are using within-spot mode, with the deconvolution information which is stored in grid.uns, as automatically determined by st.tl.cci.grid. The permutation is performed to permute the cell proportions associated with each spot, to determine spots located in LR hotspots that have certain cell types over-represented.

```python
stlearn.tl.cci.run_cci(
    grid,
    'leiden',  # Cell information either in data.obs or data.uns
    lr_results,
    min_spots=2,  # Minimum number of spots for LR to be tested.
    spot_mixtures=True,  # If True will use the deconvolution data,
    # so spots can have multiple cell types if score>cell_prop_cutoff
    cell_prop_cutoff=0.1,  # Spot considered to have cell type if score>0.1
    sig_spots=True,  # Only consider neighbourhoods of spots which had significant LR scores.
    n_perms=100,  # Permutations of cells to get background, recommend ~1000
    random_state=seed,
    n_cpus=n_cpus,
    )
```

## Diagnostic: checking for interaction and cell type frequency correlation

Should be little to no correlation; indicating the permutation has adequately controlled for cell type frequency.

```python
stlearn.pl.cci_check(grid, 'leiden', lr_results, figsize=(16, 5))
```
